### 7. In `./data/ames_prices.csv`, create an indicator that equals 1 when `Central.Air == 'Y'` and 0 otherwise.
   
   a. Estimate the proportion of homes with central air.

```python
ames_prices = pd.read_csv("../data/ames_prices.csv")

ames_prices["central_air"] = (ames_prices["Central.Air"] == "Y").astype(int)

p_hat = ames_prices["central_air"].mean()
print(p_hat)
```

   b. Using a reproducible random-number seed, generate at least 2,000 bootstrap estimates of this proportion. Report the seed and number of bootstrap replications.

```python
SEED = 5030
B = 2001
n = len(ames_prices)

rng = np.random.default_rng(SEED)
boot = np.empty(B)

for b in range(B):
    idx = rng.integers(0, n, n)
    boot[b] = ames_prices["central_air"].iloc[idx].mean()
```

c. Compute the bootstrap standard error as the sample standard deviation of the bootstrap estimates (using `ddof=1` ensures that there's the $1/(N-1)$ degrees of freedom correction).

```python
se = boot.std(ddof=1)
se
```

d. Using the original estimate and the bootstrap standard error, construct Normal 90%, 95%, and 99% confidence intervals.

```python
values = {0.10: 1.645, 0.05: 1.96, 0.01: 2.576}
for alpha, z in values.items():
    lower = p_hat - z * se
    upper = p_hat + z * se
    print(f"{100*(1-alpha):.0f}% CI: ({lower:.3f}, {upper:.3f})")
```

- Use each confidence interval to evaluate $H_0:p=.90$. At the matching significance levels $\alpha=.10,.05,.01$, state whether you reject or fail to reject the null hypothesis. Briefly interpret your results.
   - 90% CI: (0.926, 0.941) - .9 is not witin the range so we reject
   - 95% CI: (0.924, 0.942) - .9 is not witin the range so we reject
   - 99% CI: (0.921, 0.945) - .9 is not witin the range so we reject
- With these results we have strong evidence to assume that the true proportion of is not 90% and is likely higher than 90% as all of our intervals were above .9

### 8. In `./data/ames_prices.csv`, estimate the Pearson correlation between `np.arcsinh(area)` and `np.arcsinh(price)`.
   
   a. Using a reproducible random-number seed, generate at least 2,000 bootstrap estimates of the correlation by resampling paired rows. Report the seed and number of bootstrap replications.

```python
pair = ames_prices[["area", "price"]].dropna()

area = np.arcsinh(pair["area"].to_numpy(dtype=float)) # arcsinh pulls in the long right tail of areas and prices
price = np.arcsinh(pair["price"].to_numpy(dtype=float))

n = len(area)
r_hat = np.corrcoef(area, price)[0, 1]
r_hat
```

```python
SEED = 5030
B = 2001
rng = np.random.default_rng(SEED)
boot = np.empty(B)
```


   
   b. Compute the bootstrap standard error as the sample standard deviation of the bootstrap estimates (using `ddof=1` ensures that there's the $1/(N-1)$ degrees of freedom correction).

```python
for b in range(B):
    idx = rng.integers(0, n, n)
    boot[b] = np.corrcoef(area[idx], price[idx])[0, 1]

se = boot.std(ddof=1)
se
```


   
   c. Using the original estimate and the bootstrap standard error, construct centered-Normal 90%, 95%, and 99% confidence intervals.

```python
for alpha, z in values.items():
    lower = r_hat - z * se
    upper = r_hat + z * se
    print(f"{100*(1-alpha):.0f}% CI: ({lower:.3f}, {upper:.3f})")
```


   
   d. Use each confidence interval to evaluate $H_0:\rho=0$. At the matching significance levels $\alpha=.10,.05,.01$, state whether you reject or fail to reject the null hypothesis. Briefly interpret the association without making a causal claim.

- 0 is well below all three intervals so we reject the null at all three levels.
- The intervals show that larger homes tend to have a greater price as the r_hat is .72 which is strong and positive. 


   
e. Briefly explain why `area` and `price` must remain paired during resampling.
- If area and price are not paired then a home with a different price will be pair with a home of a different area. So if we change a house of a given area's price then the bootstrap will not be accurate.

```python
nhanes = pd.read_csv("../data/nhanes_data_17_18.csv", low_memory=False)

age = nhanes["AgeInYearsAtScreening"] >= 18
sleep = nhanes["SleepHoursWeekdaysOrWorkdays"].notna()

sleep = nhanes.loc[(age) & (sleep),"SleepHoursWeekdaysOrWorkdays",].to_numpy(dtype=float)

n = len(sleep)
mean_hat = sleep.mean()

SEED = 5030
B = 2001
rng = np.random.default_rng(SEED)
boot = np.empty(B)
for b in range(B):
    idx = rng.integers(0, n, n)
    boot[b] = sleep[idx].mean()

print(n, mean_hat)
```


   
   b. Compute the bootstrap standard error as the sample standard deviation of the bootstrap estimates (`ddof=1`).

```python
se = boot.std(ddof=1)
se
```


   
   c. Using the original estimate and the bootstrap standard error, construct centered-Normal 90%, 95%, and 99% confidence intervals.

```python
for alpha, z in values.items():
    lower = mean_hat - z * se
    upper = mean_hat + z * se
    print(f"{100*(1-alpha):.0f}% CI: ({lower:.3f}, {upper:.3f})")
```
